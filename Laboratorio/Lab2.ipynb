# Problema: relizar un sistema que identifique si una planta necesita riego para activar el sensor.
# Nivel de humedad suelo
# Si el nivel de humedad es menor al 40% la planta necesita riego. Si el nivel de humedad es igual o mayor al 40%, la planta no necesita riego.

```python
print("Sistema de riego inteligente")
humedad = float(input("Escribe el nivel de humedad del suelo (%): "))
```

```python
print("Datos almacenados")
print("Nivel de humedad:", humedad, "%")

#Reglas
if humedad >= 40 and humedad <= 100:
    print("La planta no necesita riego, el sensor permanece apagado")
elif humedad <40 and humedad >= 0:
    print("La planta necesita riego, se activa el sensor")

else:
    print ("Valor invalido")
```
