```python
# Laboratorio 6: Grafos
# Descripcion: Navegacion autonoma de un robot
# Situacion: Un robot movil debe desplazarse dentro de una planta industrial desde una estacion de carga hasta una estacion de trabajo.
```

```python
# Importar las librerias

#Libreria utilizada para graficar
import matplotlib.pyplot as plt
#Libreria que me permite trabajar con estructuras complejas como grafos
import networkx as nx
```

```python
# Creamos una instancia de la clase Graph
# Graph es una coleccion de nodos
 
grafo = nx.Graph()
```

```python
# Creamos el grafo utilizando el metodo add_edges
grafo.add_edges_from(
    [
        ("A","B"),
        ("A","C"),
        ("B","D"),
        ("B","D"),
        ("D","F"),
        ("F","H"),
        ("H","I"),
        ("I","J"),
        ("C","G"),
        ("G","E"),
        ("E","F"),
        ("G","J"),
    ]
)
```

```python
# Mostrar el grafo
nx.draw(
    grafo,
    with_labels = True,
    node_size= 1000,
    node_color = "#53d0fb"
)
plt.show()
```

```python
# Mostrar las rutas posibles del grafo
# Source: De donde inicia (estacion de carga)
# Target: El objetivo (estacion de trabajo)
rutas = list(
    nx.all_simple_paths(grafo, source="A", target="J")
)
```

```python
#Imprimir las rutas
print("Rutas encontradas: ")
for i, ruta in enumerate(rutas, start=1):
    print(f"Ruta {i}: {ruta}")
 
# Mostrar el total de pasos de cada ruta
pasos1 = len(rutas[0])-1
print("Pasos de la ruta 1: ", pasos1)
pasos2 = len(rutas[1])-1
print("Pasos de la ruta 2: ", pasos2)
pasos3 = len(rutas[2])-1
print("Pasos de la ruta 3: ", pasos3)
pasos4 = len(rutas[3])-1
print("Pasos de la ruta 4: ", pasos4)
```

```python
# Aplicamos la logica proposicional
P = pasos1 < pasos2 and pasos1 < pasos3 and pasos1 < pasos4 # P: La mejor ruta es la 1
Q = pasos2 < pasos1 and pasos2 < pasos3 and pasos2 < pasos4 # Q: La mejor ruta es la 2
R = pasos3 < pasos1 and pasos3 < pasos2 and pasos3 < pasos4 # R: La mejor ruta es la 3
S = pasos4 < pasos1 and pasos4 < pasos2 and pasos4 < pasos3 # S: La mejor ruta es la 4

if P:
    print("P = True")
    print("La ruta 1 es mejor opcion")
elif Q:
    print("Q = True")
    print("La ruta 2 es mejor opcion.")
elif R:
    print("R = True")
    print("La ruta 3 es mejor opcion.")
elif S:
    print("S = True")
    print("La ruta 4 es mejor opcion.")
else:
    print("Todas las condiciones son False")
    print("Hay un empate entre algunas rutas, checa los pasos de cada una")
```

```python
# Situacion: Se bloquea ahora el camino G - J
grafo.remove_edge("G","J")
```
