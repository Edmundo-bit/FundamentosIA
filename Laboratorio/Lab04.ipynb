# Proposiciones.
P: El servidor principal no disponible.
Q: La latencia es alta.
R: La perdida es superarior al limite.
S: Trafico alto.
T: Respaldo disponible.
U: Monitoreo activo.

```python
#Definicion de Variables
servidor = False # No esta disponible
latencia = True # Si supera el limite
perdidaPaquetes = True # Si supera el limite
traficoRed = True # Si es alto 
respaldoServidor = False # Si hay respaldo disponible
monitoreo = True # Si hay sistema de monitoreo activo
```

### Reglas
Existe una condicion critica cuando:
- El servidor principal no esta disponible Y
- existe latencia alta Y
- existe perdida de paquetes alta Y
- el trafico es anormalmente alto Y
- el servidor de respaldo no esta disponible Y
- el sistema de monitoreo esta activo.

```python
condicionCritica = ( servidor == False and latencia == True and perdidaPaquetes == True and
                     traficoRed == True and respaldoServidor == False and monitoreo == True)
```

```python
print ("¿El servidor principal esta disponible?", condicionCritica)
```
